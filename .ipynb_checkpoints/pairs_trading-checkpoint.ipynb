```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
import scipy
from datetime import datetime
import pytz
```

```python
pair = ['GS','JPM']
```

```python
data = yf.download(pair,start=datetime(2020,1,1,tzinfo=pytz.utc), end=datetime(2026,1,1,tzinfo=pytz.utc),multi_level_index=False,auto_adjust=False)['Adj Close']
data = data.rename(columns ={'GS':'stockA',
                    'JPM':'stockB'})
```

```python
data['spread'] = data['stockA'] / data['stockB']
data
```

```python
data['z_score'] = scipy.stats.zscore(data['spread'],nan_policy='raise')
data
```
